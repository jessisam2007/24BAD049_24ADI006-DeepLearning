# EXPT NO: 7 — TEXT PREPROCESSING AND EMBEDDING GENERATION USING NLP LIBRARIES

**Roll No:** 24BAD049

**Aim:** To perform text preprocessing using NLTK and generate word embeddings using Gensim Word2Vec to analyze semantic relationships between words.

## PART A: DATASET PREPARATION AND TEXT PREPROCESSING

```python
!pip install -q gensim

import string
import nltk
from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

ROLL_NO = "24BAD049"
print("Roll No:", ROLL_NO)

for res in ["brown", "punkt", "punkt_tab", "stopwords", "wordnet", "omw-1.4"]:
    nltk.download(res, quiet=True)

category = "news"
brown_sents = brown.sents(categories=category)
print("Category selected:", category)
print("Total sentences :", len(brown_sents))

print("\n--- Sample sentences from the dataset ---")
for i in range(3):
    print(f"Sentence {i+1}: {' '.join(brown_sents[i])}")

original_sentences = [" ".join(sent) for sent in brown_sents]

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

preprocessed_sentences = []
for sentence in original_sentences:
    sentence = sentence.lower()

    tokens = word_tokenize(sentence)

    tokens = [t for t in tokens if t.isalpha()]
    tokens = [t for t in tokens if t not in stop_words]

    tokens = [lemmatizer.lemmatize(t) for t in tokens]

    preprocessed_sentences.append(tokens)

print("\n--- Original vs Preprocessed text ---")
for i in range(3):
    print(f"\nOriginal     [{i+1}]: {original_sentences[i]}")
    print(f"Preprocessed [{i+1}]: {' '.join(preprocessed_sentences[i])}")
```

## PART B: VOCABULARY GENERATION AND WORD FREQUENCY ANALYSIS

```python
from nltk import FreqDist

print("Roll No:", ROLL_NO)

all_words = [word for sentence in preprocessed_sentences for word in sentence]

vocabulary = sorted(set(all_words))

freq_dist = FreqDist(all_words)

print("Total words after preprocessing:", len(all_words))
print("Vocabulary size                :", len(vocabulary))
print("Frequency distribution         :", freq_dist)

print("\n--- Ten most frequent words ---")
print(f"{'Rank':<6}{'Word':<15}{'Frequency'}")
for rank, (word, count) in enumerate(freq_dist.most_common(10), start=1):
    print(f"{rank:<6}{word:<15}{count}")

w2v_sentences = [sentence for sentence in preprocessed_sentences if len(sentence) > 0]
print("\nSentences prepared for Word2Vec:", len(w2v_sentences))
print("Sample tokenized sentence      :", w2v_sentences[0])
```

## PART C: WORD EMBEDDING GENERATION USING WORD2VEC

```python
from gensim.models import Word2Vec

print("Roll No:", ROLL_NO)

model = Word2Vec(
    sentences=w2v_sentences,
    vector_size=100,
    window=5,
    sg=1,
    min_count=2,
    workers=1,
    epochs=30,
    seed=42
)
print("Word2Vec model trained. Vocabulary size:", len(model.wv))

selected_words = ["president", "government", "school"]
for word in selected_words:
    vector = model.wv[word]
    print(f"\nWord      : {word}")
    print(f"Dimensions: {vector.shape}")
    print(f"Vector    : {vector}")

target_word = "president"
print(f"\n--- Five most similar words to '{target_word}' ---")
for word, score in model.wv.most_similar(target_word, topn=5):
    print(f"{word:<15}{score:.4f}")
```

## PART D: WORD SIMILARITY ANALYSIS AND EXPERIMENT MANAGEMENT

```python
print("Roll No:", ROLL_NO)

word_pairs = [("president", "government"), ("city", "state"), ("game", "team")]

similarity_scores = {}
print("--- Cosine similarity of word pairs ---")
for w1, w2 in word_pairs:
    score = model.wv.similarity(w1, w2)
    similarity_scores[(w1, w2)] = score
    print(f"{w1:<12}{w2:<12}{score:.4f}")

best_pair = max(similarity_scores, key=similarity_scores.get)
print(f"\nMost similar pair: {best_pair} (score = {similarity_scores[best_pair]:.4f})")

print("\n--- Observation ---")
print(f"The pair {best_pair} has the highest cosine similarity, which means these words")
print("appear in more similar contexts in the Brown 'news' text. Skip-gram learns each word's")
print("vector from the words surrounding it, so words sharing similar neighbours get vectors")
print("that point in similar directions in the 100-dimensional space (cosine score closer to 1).")
```
